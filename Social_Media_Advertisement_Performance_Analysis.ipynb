```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

```python
ad_summary = df.groupby('Ads').agg({
    'Impressions': 'sum',
    'Reach': 'sum',
    'Link clicks': 'sum',
    'Amount spent (INR)': 'sum',
    'Meta leads': 'sum',
    'Results': 'sum'
}).reset_index()

ad_summary.head()
```

```python
df[['Ads', 'Impressions', 'Reach', 'Link clicks',
    'Amount spent (INR)', 'Meta leads', 'Results']].to_string()
```

```python
ad_summary = df.groupby('Ads').agg({
    'Link clicks': 'sum',
    'Meta leads': 'sum',
    'Results': 'sum'
}).reset_index()

ad_summary
```

```python
ad_summary['Total Actions'] = (
    ad_summary['Link clicks'].fillna(0) +
    ad_summary['Meta leads'].fillna(0)
)

ad_summary
```

```python
ranking = ad_summary.sort_values(
    'Total Actions',
    ascending=False
)

ranking
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.bar(
    ranking['Ads'],
    ranking['Total Actions']
)

plt.xlabel('Advertisement')
plt.ylabel('Total Actions')
plt.title('Advertisement Performance Comparison')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
x = np.arange(len(ranking))
width = 0.35

plt.figure(figsize=(10,6))

plt.bar(
    x - width/2,
    ranking['Link clicks'].fillna(0),
    width,
    label='Link Clicks'
)

plt.bar(
    x + width/2,
    ranking['Meta leads'].fillna(0),
    width,
    label='Meta Leads'
)

plt.xlabel('Advertisement')
plt.ylabel('Count')
plt.title('Link Clicks vs Meta Leads by Advertisement')

plt.xticks(x, ranking['Ads'], rotation=45, ha='right')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
ad_summary['Lead Conversion Rate'] = np.where(
    ad_summary['Link clicks'] > 0,
    ad_summary['Meta leads'] /
    ad_summary['Link clicks'] * 100,
    np.nan
)

ad_summary
```

```python
ad_summary.sort_values(
    'Lead Conversion Rate',
    ascending=False
)
```

```python
print("TOP ADS")
print("=" * 40)

print("\nMost Link Clicks:")
print(
    ad_summary.loc[
        ad_summary['Link clicks'].idxmax(),
        ['Ads', 'Link clicks']
    ]
)

print("\nMost Meta Leads:")
print(
    ad_summary.loc[
        ad_summary['Meta leads'].idxmax(),
        ['Ads', 'Meta leads']
    ]
)

print("\nMost Total Actions:")
print(
    ad_summary.loc[
        ad_summary['Total Actions'].idxmax(),
        ['Ads', 'Total Actions']
    ]
)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14,9))
axes = axes.flatten()

for i, row in ad_summary.iterrows():

    ax = axes[i]

    clicks = row['Link clicks']
    leads = row['Meta leads']

    ax.bar(['Clicks', 'Leads'], [clicks, leads])

    ax.set_title(row['Ads'], fontsize=12, fontweight='bold')
    ax.set_ylabel('Count')

    ax.text(0, clicks + 1, str(int(clicks)), ha='center', fontweight='bold')
    ax.text(1, leads + 1, str(int(leads)), ha='center', fontweight='bold')

    ax.spines[['top','right']].set_visible(False)

plt.suptitle(
    '📊 Advertisement Performance Scorecards',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

ranking = ad_summary.sort_values('Meta leads', ascending=True)

plt.figure(figsize=(11,6))

bars = plt.barh(
    ranking['Ads'],
    ranking['Meta leads']
)

plt.xlabel('Meta Leads')
plt.title('🏆 Ad Performance Leaderboard', fontsize=16, fontweight='bold')

for bar in bars:
    plt.text(
        bar.get_width() + 0.15,
        bar.get_y() + bar.get_height()/2,
        int(bar.get_width()),
        va='center',
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

matrix = ad_summary[['Ads', 'Link clicks', 'Meta leads']].copy()

matrix['Performance Score'] = (
    matrix['Link clicks'].rank(pct=True) * 0.4 +
    matrix['Meta leads'].rank(pct=True) * 0.6
)

matrix = matrix.sort_values('Performance Score', ascending=False)

plt.figure(figsize=(12,7))

plt.scatter(
    matrix['Link clicks'],
    matrix['Meta leads'],
    s=500,
    alpha=0.7
)

for _, row in matrix.iterrows():
    plt.annotate(
        row['Ads'],
        (row['Link clicks'], row['Meta leads']),
        xytext=(8,8),
        textcoords='offset points',
        fontsize=10,
        fontweight='bold'
    )

plt.xlabel('Link Clicks')
plt.ylabel('Meta Leads')
plt.title('🎯 Campaign Performance Matrix', fontsize=17, fontweight='bold')
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
top_clicks = ad_summary.loc[ad_summary['Link clicks'].idxmax()]
top_leads = ad_summary.loc[ad_summary['Meta leads'].idxmax()]

fig, ax = plt.subplots(figsize=(12,5))
ax.axis('off')

ax.text(0.05, 0.75, '🏆 TOP PERFORMING ADS',
        fontsize=22, fontweight='bold')

ax.text(0.05, 0.48,
        f"🚀 Traffic Leader\n{top_clicks['Ads']}\n{int(top_clicks['Link clicks'])} Link Clicks",
        fontsize=15)

ax.text(0.55, 0.48,
        f"🎯 Lead Generation Leader\n{top_leads['Ads']}\n{int(top_leads['Meta leads'])} Meta Leads",
        fontsize=15)

ax.text(0.05, 0.12,
        'Insight: Evaluate ads separately for traffic generation and lead generation.',
        fontsize=12)

plt.tight_layout()
plt.show()
```
